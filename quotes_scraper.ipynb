## Libraries

```python
# import libraries
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import csv
```

## Scraping Function

```python
# create separate lists
quote_list=[]
author_list=[]
tag_list=[]

# create a function to find the data
def find_quotes(url):
  # run while it have another 'next' url
  while url:
    response = requests.get(url)
    soup = BeautifulSoup(response.text, "html.parser")
    # find all quotes
    quotes=soup.find_all('div',class_='quote')
    # for each quote
    for quote in quotes:
      tags=quote.find_all('a',class_='tag')
      # create a list for the tags of each quote
      each_tag_list=[]
      for tag in tags:
        # store all the tags of a quote in this list
        each_tag_list.append(tag.text)
        # append this tag list in the main tag list
      tag_list.append(each_tag_list)

       # store the quote in the list
      single_quote=quote.find_all('span')
      quote=(single_quote[0].text)
      quote_list.append(quote)
      # store the author in the list
      author=(single_quote[1].find('small').text)
      author_list.append(author)

      # find next button
    next_button = (soup.find('li',class_='next'))
    if next_button:
      # if available
      next_link=(next_button.find('a'))
      # combine its url with original url and then again use it
      url=urljoin(url,next_link['href'])
    else:
      url=None
      # return all lists
  return quote_list,author_list,tag_list
```

## Run Scraper

```python
# run the function
quote_list,author_list,tag_list=find_quotes("https://quotes.toscrape.com/")
```

## Save Data to CSV

```python
# store data in the csv file 'quote.csv'

with open("quotes.csv", "w", newline="", encoding="utf-8") as file:

    writer = csv.writer(file)

    # Header rpw
    writer.writerow(["Quote", "Author", "Tags"])

    # Data rows
    for quote, author, tags in zip(quote_list, author_list, tag_list):
      # open the tag list
        writer.writerow([quote, author, ", ".join(tags)])
```
